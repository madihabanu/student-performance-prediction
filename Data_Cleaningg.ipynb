```python

import pandas as pd
from pathlib import Path

# Find the Excel files
data_folder = Path.cwd()

excel_files = list(data_folder.glob("*.xlsx"))

# If files are in the parent folder
if not excel_files:
    data_folder = data_folder.parent
    excel_files = list(data_folder.glob("*.xlsx"))

print("DATASET INSPECTION")
print("=" * 50)

for file in excel_files:
    print(f"\nFile: {file.name}")

    excel = pd.ExcelFile(file)

    print("Number of sheets:", len(excel.sheet_names))
    print("Sheet names:", excel.sheet_names)

    for sheet in excel.sheet_names:
        df = pd.read_excel(file, sheet_name=sheet, header=None)

        print(
            f"  {sheet}: "
            f"Rows = {df.shape[0]}, "
            f"Columns = {df.shape[1]}"
        )
```

```python

import pandas as pd
from pathlib import Path

# Find the Excel files
folder = Path.cwd()

files = list(folder.glob("*.xlsx"))

if not files:
    folder = folder.parent
    files = list(folder.glob("*.xlsx"))

# Select the first Excel file
file_path = files[0]

# Read the consolidated sheet
df = pd.read_excel(
    file_path,
    sheet_name="Consolidated",
    header=None
)

# Display first 20 rows
print("First 20 rows:")
display(df.head(20))
```

```python

# Find exact duplicate rows
duplicate_rows = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_rows)

# Display duplicate rows
display(df[df.duplicated(keep=False)])
```

```python

missing_percentage = (
    df.isnull().mean() * 100
).round(2)

print("Missing values percentage:")
display(missing_percentage)
```
